```python
def compute_mean(numbers):
    """
    Compute_mean is a function that generates and prints the mean of a list of numbers.

    Parameters
    -----------
    numbers : int or float (List of numbers for the function)

    Examples
    ---------
    >>>> compute_mean([3,6,9)]
    6.0
    """

    mean_value = sum(numbers) / len(numbers)
    print(mean_value)
```

```python
compute_mean([2, 4, 6, 8])
```
